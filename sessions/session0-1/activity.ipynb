# Dynamics of the drone

+ We control the vertical movement of the drone only. 
+ The vertical acceleration is influenced by three forces: 
    + the weight of the drone,
    + air resistance proportional to its speed,
    + upward thrust produced by its propellers. 
+ The controller function $C$ determines the upward thrust $\tau(t)$ (at time $t$) as a function of the meausred height $h(t)$, the vertical speed $v(t)$, and the desired hovering height $\bar{h}$.
\begin{equation*}
\tau(t) = C(h(t),v(t),\bar{h} )
\end{equation*}
+ The thrust is clipped to remain within $[0,\bar{\tau}]$:
\begin{equation*}
\tau_c = \max( 0 , \min( \tau , \bar{\tau} ) )
\end{equation*}
+ Differential equation for the position and speed:
\begin{align*}   
m \dot{v} &= \tau - \nu \: v - m g \\
\dot{h} &= v
\end{align*}
+ $m$ is mass, $g$ is gravity, $\nu$ is a friction coefficient.
