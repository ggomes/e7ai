# AI-assisted implementation of a controller for a hovering drone 

+ In this activity we will work directly with an AI code assistant (Copilot) to create a controller for a hovering drone. 
+ The idea is to *not* write any of the code by hand, but instead to instruct the AI to do it. 
+ Dynamical system. We will control the vertical movement of the drone only, no steering. This verticl movement is affected by three forces: 
    + the weight of the drone,
    + air resistance proportional to its speed,
    + upward thrust produced by its propellers. 
+ The "controller function" $C$ determines the upward thrust $\tau$ as a function  the meausred height $h(t)$ and vertical speed $v(t)$ of the drone, as well as the desired hovering height $\bar{h}$.
\begin{equation*}
\tau = C(h(t),v(t),\bar{h} )
\end{equation*}
+ $t$ is the current time.

+ Differential equation for the position and speed:
\begin{align*}   
m \dot{v}(t) &= \tau - \nu \: v(t) - m g \\
\dot{h}(t) &= v(t) 
\end{align*}
+ $nu$ is related to the viscous friction coefficient of the air.
