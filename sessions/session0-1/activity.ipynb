Motors provide upward thrust only: zero voltage means free fall.

## Engineering requirement

With the default initial conditions (height 0 m, velocity 0 m/s) and target height 10 m, the controller must:

- bring the drone within 0.15 m of 10 m;
- settle by step 70;
- remain within the tolerance through step 100.

The program should print a pass/fail result and save `drone_response.png`, showing the response, target, tolerance band, and settling deadline. Maximum voltage until the target followed by zero voltage is not sufficient: the drone has inertia, so the controller must use velocity feedback to brake and must provide enough voltage to hover against gravity and damping.
