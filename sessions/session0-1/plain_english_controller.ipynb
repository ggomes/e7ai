# Plain-English Logic Prompts  
## The Natural Language Controller

### Purpose

In this activity, you will use GitHub Copilot to help translate an engineering control policy into Python code.

You will **not** implement the tank model. The tank model is provided as a black-box module. Your responsibility is to describe the control policy, evaluate Copilot’s suggestion, and test the resulting controller.

### Learning objectives

By completing this activity, you should be able to:

- describe an engineering rule precisely in plain English;
- recognize the steps in a simulation loop;
- use Copilot to assist with Python syntax;
- check whether generated code matches your intended behavior;
- test a controller using multiple initial conditions.

### Files

Open the `session0-1` folder in VS Code.

- `tank_system.py` contains the provided tank model. **Do not modify this file.**
- `student_controller.py` contains the controller you will complete.

The tank module provides:

```python
tank_system.set_valve_position(current_level, target_level)
```

This returns a valve command between `-100` and `100`.

- Negative values drain the tank.
- Positive values fill the tank.
- Zero leaves the valves closed.

The module also provides:

```python
tank_system.advance(current_level, valve_position)
```

This advances the simulation by one time step.

### Instructions

1. Open `student_controller.py`.

2. Read the existing comments and code. Identify the four steps repeated by the loop:

   1. Observe the tank level.
   2. Decide what control action is appropriate.
   3. Set the valve position.
   4. Advance the tank model.

3. Write your control policy in plain English comments. Your policy must include rules for:

   - the level being substantially above the target;
   - the level being substantially below the target;
   - the level being close to the target.

   For example:

   ```python
   # If the water level is more than 10% above the target, drain the tank.
   # If the water level is more than 10% below the target, fill the tank.
   # Otherwise, keep the level near the target.
   ```

4. Place the cursor in the control-flow section and ask Copilot to complete or expand the `if`/`elif`/`else` structure. Press **Tab** to accept a suggestion only after reading it.

5. Review the generated code carefully. Check:

   - Are the “above” and “below” conditions correct?
   - Are the threshold values correct?
   - Does the code match your English comments?
   - Does it use the supplied tank module correctly?
   - Does it behave sensibly near the target?

6. Run the program. Observe the printed tank levels and valve commands.

7. Test at least these cases:

   - initial level below the target;
   - initial level above the target;
   - initial level close to the target.

8. Revise your code if the behavior does not match your policy. You may ask Copilot for help, but you remain responsible for the final code.

9. Add a short explanation at the bottom of `student_controller.py` answering:

   - What did Copilot suggest?
   - What did you accept, reject, or change?
   - How did you verify that the controller followed your rules?
   - What limitation or assumption did you notice?

### Important rule

Copilot is an assistant, not an authority. You are responsible for understanding, testing, and explaining the code it generates. Do not modify the implementation of `tank_system.py`.
