# Plain-English Logic Prompts  
## The Natural Language Controller: Hovering Drone

### Purpose

In this activity, you will use GitHub Copilot to help translate an engineering control policy into Python code.

You will **not** implement the drone model. The drone model is provided as a black-box module. Your responsibility is to describe the control policy, evaluate Copilot's suggestion, and test the resulting controller.

### Learning objectives

By completing this activity, you should be able to:

- describe an engineering rule precisely in plain English;
- recognize the observe, decide, actuate, and advance steps in a simulation loop;
- understand why position and velocity feedback matter for a system with inertia;
- use Copilot to assist with Python syntax without delegating engineering decisions;
- check whether generated code matches your intended behavior;
- test a controller using multiple initial conditions and explain its limitations.

### Files

Open the `session0-1drone` folder in VS Code.

- `drone_system.py` contains the provided drone model. **Do not modify this file.**
- `student_controller.py` contains the controller you will complete.

Read the docstrings in `drone_system.py`. The model provides:

```python
drone_system.V_MAX
drone_system.advance(current_position, current_velocity, voltage)
```

`V_MAX` is the largest allowed motor voltage. `advance` moves the simulation forward by one time step and returns the new position and velocity.

The drone moves only vertically. Its acceleration is determined by upward propeller thrust, downward gravity, and viscous damping that opposes velocity. The propellers cannot push downward, so a voltage of zero produces free fall.

### Engineering specification

The drone starts at rest on the floor at height 0 m, and the target height is 10 m. Each loop iteration represents 0.1 seconds. Your controller must:

- bring the drone within 0.15 m of the target;
- settle by the end of step 70;
- remain within that tolerance through step 100.

The program reports whether the specification was met and produces `drone_response.png`. The plot shows the target height, the allowed tolerance band, and the simulated response.

The specification is intentionally not satisfied by applying maximum voltage until the target is reached and then setting voltage to zero. The drone has inertia and will continue moving after the voltage changes. You must use velocity feedback to brake while approaching the target and apply enough voltage near the target to balance gravity and damping. Do not change `drone_system.py` to make the test pass.

### Instructions

1. Open `student_controller.py`.

2. Read the existing comments and code. Identify the four steps repeated by the loop:

   1. Observe the drone's position and velocity.
   2. Decide what control action is appropriate.
   3. Set the motor voltage.
   4. Advance the drone model.

3. Write your control policy in plain-English comments. Your policy must include rules for:

   - the drone being substantially above the target;
   - the drone being substantially below the target;
   - upward velocity while the drone is approaching the target;
   - hovering near the target despite gravity;
   - clamping the command to the range `0` to `drone_system.V_MAX`.

   For example:

   ```python
   # If the drone is below the target, use upward thrust, but reduce it
   # when upward velocity is carrying the drone toward the target.
   # If it is above the target or moving upward too quickly, reduce thrust
   # to brake without commanding a negative voltage.
   # Near the target, use enough voltage to counter gravity and damping.
   ```

   These comments are a starting point, not a complete controller. Choose and explain your own thresholds and gains.

4. Place the cursor in the control-flow section and ask Copilot to complete or expand the `if`/`elif`/`else` structure. Press **Tab** to accept a suggestion only after reading it.

5. Review the generated code carefully. Check:

   - Does it use both position error and velocity feedback?
   - Does it reduce thrust early enough to brake before overshooting?
   - Does it provide enough thrust to hover near the target rather than free-fall?
   - Are the thresholds and signs correct?
   - Can the voltage ever be negative or exceed `drone_system.V_MAX`?
   - Does the code match your English comments and use the supplied model correctly?

6. Run the program. Observe the printed positions, velocities, and voltages. Read the pass/fail report and inspect `drone_response.png`.

7. Test at least these cases by changing the initial conditions or target and running the program again:

   - starting below the target at rest;
   - starting above the target;
   - starting with downward velocity;
   - starting close to the target with a nonzero upward velocity.

   Restore the original specification values before your final run.

8. Revise your code if the behavior does not match your policy. You may ask Copilot for help, but you remain responsible for the final code. Do not change the drone model to make the test pass.

9. Add a short explanation at the bottom of `student_controller.py` answering:

   - What did Copilot suggest?
   - What did you accept, reject, or change?
   - How did position and velocity feedback affect the result?
   - How did you verify the voltage limits and settling requirement?
   - What limitation or assumption did you notice?
